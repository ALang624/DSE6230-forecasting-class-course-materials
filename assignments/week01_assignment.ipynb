```python
# QUESTION 2.6

# AUS ARRIVALS SETUP
aus_arrivals = pd.read_csv('aus_arrivals.csv')
aus_arrivals['Quarter'] = pd.to_datetime(aus_arrivals['Quarter'])

# List of the visitor home countries.
countries = ['Japan', 'NZ', 'UK', 'US']
arrivals_subset = aus_arrivals[aus_arrivals['Origin'].isin(countries)]

# Plot by Series: For each quarter, plot the arrivals by origin.
fig, ax = plt.subplots(figsize=(10, 6))
for origin, group in arrivals_subset.groupby('Origin'):
    ax.plot(group['Quarter'], group['Arrivals'], label=origin)
ax.set_xlabel('Quarter')
ax.set_ylabel('Arrivals')
ax.legend()
plt.show()

  # ANALYSIS
  # Overall, the vast majority of visitors to Australia comes from New Zealand. They are the only group to surpass 300,000 arrivals
  # amongst our visiting quarters. The UK comes close, with a steadily increasing number of arrivals each year. Japan was consistently
  # growing until roughly 1996, at which point it plateaus and eventually begins to decrease by the 2010s. Lastly, the US remains a
  # gradually increasing number, beginnign low and remaining around 100,000 throughout.
```

```python
# Seasonal Plot: Overlay each year's pattern on top of each other, one line per year.
for origin, group in arrivals_subset.groupby('Origin'):
    g = group.set_index('Quarter')['Arrivals']
    pivot = g.groupby([g.index.year, g.index.quarter]).mean().unstack(level=0)
    # Arranging the squares
    fig, ax = plt.subplots(figsize=(8, 4))
    pivot.plot(ax=ax, legend=False) # Legend is removed because it is illegible otherwise.
    ax.set_title(f'Seasonal Plot: {origin}')
    ax.set_xlabel('Time of Year')
    plt.show()

# Subseries Plot: One small panel per quarter, showing that quarter's value across years.
for origin, group in arrivals_subset.groupby('Origin'):
    g = group.set_index('Quarter')['Arrivals']
    # Arranging the squares
    fig, axes = plt.subplots(1, 4, figsize=(12, 3), sharey=True)
    for q, ax in zip([1, 2, 3, 4], axes):
        qdata = g[g.index.quarter == q]
        ax.plot(qdata.index.year, qdata.values)
        ax.set_title(f'Q{q}')
    fig.suptitle(f'Subseries Plot: {origin}')
    plt.tight_layout()
    plt.show()

# ANALYSIS
# Amongst visitors, it seems the most popular time between all nations is during the summer, in which all groups but the UK
# take a sharply increased turn towards visitation. The UK and Japan tends to visit most during the winter months, while NZ
# tends to avoid the colder months. Lastly, the US remains largely scattered, visiting inconsistently.
```

```python
# QUESTION 2.7

# IMPORTS
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.graphics.tsaplots import plot_acf

# AUS RETAIL SETUP
aus_retail = pd.read_csv('aus_retail.csv')
aus_retail['Month'] = pd.to_datetime(aus_retail['Month'])

# Seeing as Victoria was popular previously, let's
# use this as our basis for seasonality.
chosen_state = 'Victoria'
chosen_industry = 'Food retailing'

# Our series of Victoria retail
retail_series = aus_retail[
    (aus_retail['State'] == chosen_state) &
    (aus_retail['Industry'] == chosen_industry)
].set_index('Month')['Turnover'].sort_index()
# Note - our X axis will display year numbers, but each point is still
# month-by-month in actuality.

# 1. Time plot
retail_series.plot(figsize=(10, 4), title='Retail Turnover')
plt.show()

# 2. Seasonal decomposition
decomposition = seasonal_decompose(retail_series, model='additive', period=12)
decomposition.plot()
plt.show()

# 3. Lag plot
pd.plotting.lag_plot(retail_series)
plt.title('Lag Plot')
plt.show()

# 4. ACF plot
plot_acf(retail_series, lags=36)
plt.show()

# 5. One Year of Seasonality
decomposition.seasonal['2015':'2016'].plot(title='One Year of Seasonality')

# ANALYSIS
# As the years move forward, turnover steadily increases, doing so
# to a greater degree as well overtime. In terms of seasonality, we can see continuous
# and massive spikes when winter holidays would be coming, as well as small rises for other holidays
# as well. Our ACF plot slowly decays over time, with some notable movement due to the spikes we experience
# with seasons, but overall not shifting greatly in value between each point, indicating a downwards trend.
```

```python
# QUESTION 2.8
# NOTE : As there are many graphs present, I'm unsure of if it is required to analyze each individual graph in full, or
# if this simply is for my own observation and reading. If desired, I'm happy to return to this section after submission and
# write up observations on each set, but am presently leaving it blank to keep clean an already massive section of graphs
# where there is no question put forward to answer.

# US EMPLOYMENT SETUP
us_employment = pd.read_csv('us_employment.csv')
us_employment['ds'] = pd.to_datetime(us_employment['ds'])

# AUS PRODUCTION SETUP
aus_production_full = pd.read_csv('aus_production.csv')
aus_production_full['ds'] = pd.to_datetime(aus_production_full['ds'])
aus_production_full = aus_production_full.set_index('ds')

# PELT SETUP
pelt_full = pd.read_csv('pelt.csv')
pelt_full['ds'] = pd.to_datetime(pelt_full['ds'])

# PBS SETUP
pbs = pd.read_csv('pbs.csv')
pbs['Month'] = pd.to_datetime(pbs['Month'].str.replace('Sept', 'Sep'), format='%Y %b')  # 'Sept' isn't a standard abbreviation

# US GASOLINE SETUP
us_gasoline = pd.read_csv('us_gasoline.csv')
us_gasoline['ds'] = pd.to_datetime(us_gasoline['ds'])


def diagnose_series(series, title, period=12, lags=36):
  # Create the initial plot.
    fig, ax = plt.subplots(figsize=(10, 4))
    series.plot(ax=ax, title=title)
    plt.show()

  # Create the decomposition plot.
    decomposition = seasonal_decompose(series, model='additive', period=period)
    decomposition.plot()
    plt.suptitle(f'{title} - Decomposition', y=1.02)
    plt.show()

  # Create the lag plot.
    pd.plotting.lag_plot(series)
    plt.title(f'{title} - Lag Plot')
    plt.show()

  # Create the ACF plot.
    plot_acf(series, lags=lags)
    plt.title(f'{title} - ACF')
    plt.show()
```

```python
# 1. Total Private employment (monthly)
total_private = us_employment[us_employment['unique_id'] == 'Total Private'].set_index('ds')['y']
diagnose_series(total_private, 'US Employment - Total Private', period=12)
```

```python
# 2. Bricks production (quarterly)
bricks = aus_production_full['Bricks'].dropna()
diagnose_series(bricks, 'Bricks Production', period=4)
```

```python
# 3. Hare pelts (annual)
hare = pelt_full[pelt_full['unique_id'] == 'hare'].set_index('ds')['y']
fig, ax = plt.subplots(figsize=(10, 4))
hare.plot(ax=ax, title='Hare Pelts')
plt.show()
pd.plotting.lag_plot(hare)
plt.title('Hare Pelts - Lag Plot')
plt.show()
plot_acf(hare, lags=30)
plt.title('Hare Pelts - ACF')
plt.show()
```

```python
# 4. PBS H02 cost (monthly)
h02 = pbs[pbs['ATC2'] == 'H02']
h02_cost = h02.groupby('Month')['Cost'].sum().sort_index()
diagnose_series(h02_cost, 'PBS H02 Cost', period=12)
```

```python
# 5. US gasoline (weekly data, period ~52)
barrels = us_gasoline.set_index('ds')['y'].sort_index()
diagnose_series(barrels, 'US Gasoline Barrels', period=52)
```

```python
# QUESTION 2.9

# AUS LIVESTOCK SETUP
aus_livestock = pd.read_csv('aus_livestock.csv')
aus_livestock['ds'] = pd.to_datetime(aus_livestock['ds'])

# Set up series of pig slaughters within Victoria from 1990-95.
pigs = aus_livestock[aus_livestock['unique_id'] == 'Victoria_Pigs'].set_index('ds')['y'].sort_index()
pigs = pigs.loc['1990':'1995']

# Plot the slaughters
fig, ax = plt.subplots(figsize=(10, 4))
pigs.plot(ax=ax, title='Victorian Pig Slaughters, 1990-1995')
plt.show()

# ACF plot for the slaughters.
plot_acf(pigs, lags=36)
plt.title('ACF - Victorian Pig Slaughters')
plt.show()

# ANALYSIS
# For our data to escape matching white noise, we need to see a great deal of movement from an ACF of 0, as well as preferably
# movement outside of our confidence band. In our case here, we see in our original series plot a number of notable spikes alongside
# a steadily increasing number of slaughters, with many sharp rises and falls throughout. Our ACF plot confirms this, showing within our
# bars a great deal of variance - our first bar reaches all the way to 1, followed by a sharp falloff to ~0.7, and then a set of steadily decreasing
# values until we hit the lag point of 27. At that point, we see *negative* movement, one that mirrors the pattern seen on the positive
# segment of our y-axis. There is, indeed, an observable trend, and a strong linear one at that - very exciting!
```

```python
# QUESTION 10

# GAFA STOCK SETUP
gafa_stock = pd.read_csv('gafa_stock.csv')
gafa_stock['ds'] = pd.to_datetime(gafa_stock['ds'])

# Retrieve GOOG from our data frame.
goog = gafa_stock[gafa_stock['unique_id'] == 'GOOG_Close'].set_index('ds')['y'].sort_index()

# Create DGOOG, the daily changes.
dgoog = goog.diff().dropna()

# Plot of DGOOG.
fig, ax = plt.subplots(figsize=(10, 4))
dgoog.plot(ax=ax, title='Daily Change in GOOG Closing Price')
plt.show()

# The ACF Plot of DGOOG.
plot_acf(dgoog, lags=36)
plt.title('ACF - Daily GOOG Price Changes')
plt.show()

# ANALYSIS
# Our plotted graph of DGOOG shows a mostly consistent level of change throughout the years, mostly contained within 30 points in the first half,
# and moving into more volatile changes in the latter half. We see two notable spikes, namely a massive change in mid-2015 - the creation of Alphabet
# parent company - and a significnatly higher set of spikes and drops through 2018, when a massive data breach occurred, as well as a walkout
# of employees due to sexual misconduct scandals. Within our ACF, we see that our data does in fact look like white noise however, with only small
# bouts of movement up or down, alongside the one spike at point 1 that represents the immense volatility found in a few specific, isolated areas.
```
